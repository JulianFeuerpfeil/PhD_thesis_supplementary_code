## 5.5.3 Numerical Experiments Related to Heuristic 5.5.2 (Fixing K and varying p)
Here we present the code to reproduce the numbers presented in Tables 5.2, 5.3, and 5.4. The cells can be evaluated independtly.

```julia
# Table 5.2 (r2=0)

fs=[
t^5 - t^4 - 34*t^3 + 285*t + 285, 
t^5 - 55*t^3 + 605*t - 22, 
t^6 - 10*t^4 + 24*t^2 - 8, 
t^7 - 7*t^5 + 13*t^3 - 5*t - 1,
t^8 - 13*t^6 + 44*t^4 - 17*t^2 + 1
]

for f in fs 
	K,_=number_field(f)
	println("\n For the number field defined by $(defining_polynomial(K)) and signature $(signature(K)) we find")
	@show rank_with_varying_prime(K,1:10^5)
end
```

```julia
# Table 5.3 (r1=1)

fs=[
t^5 - 41*(t - 1)*(t - 2)*(t - 3),
t^6 - t^5 - t^4 - 2*t^3 + 2*t^2 + 3*t - 1,
t^7 - t^5 - 4*t^3 + 4*t - 1 ,
t^8 - 6*t^4 + 5*t^2 - 1 ,
t^8 - 4*t^6 + t^4 + 5*t^2 - 2
]

for f in fs 
	K,_=number_field(f)
	println("\n For the number field defined by $(defining_polynomial(K)) and signature $(signature(K)) we find")
	@show rank_with_varying_prime(K,1:10^5)
end
```

```julia
# The results for Q(sqrt[4]{11}) are not included in the tables, but are computed as follows:

K,_=number_field(t^4-11)

println("\n For the number field defined by $(defining_polynomial(K)) and signature $(signature(K)) we find")
@show rank_with_varying_prime(K,1:10^7)
```

```julia
# Table 5.5 (r1>=2)

fs=[
t^6-t^5-t^4+2*t^3+2*t^2+3*t-1,
t^7 - t^4 - t^3 + t^2 + 1, 
t^8 + t^6 - t^3 - t^2 - 1, 
t^8 - t^7 - t^3 + 2*t^2 - t + 1, 
t^9 -t^7 -t^6 +t^4 -t^3 +t+1
]

for f in fs 
	K,_=number_field(f)
	println("\n For the number field defined by $(defining_polynomial(K)) and signature $(signature(K)) we find")
	@show rank_with_varying_prime(K,1:10^5)
end
```

## 5.5.3 Numerical Experiments Related to Heuristic 5.5.2 (Fixing p and varying K)
Here we present the code to reproduce the numbers presented in Tables 5.5, 5.6, and 5.7. Please first execute the following cell to define the polynomial ring in two variables. The rest of the cells can be evaluated independently. 

```julia
QXt,(X,t)=polynomial_ring(QQ,["X","t"])
```

```julia
# Table 5.5 (X^3+t*X^2+(t-3)*X-1 with generic Galois group C_3)

g=X^3+t*X^2+(t-3)*X-1

ps=[3,5,7,11,31]

for p in ps 
	println("\n For g=$(g) in the range [-1000,1000] the output of the function rank_with_varying_field for p=$(p) gives the following output:")
	@show rank_with_varying_field(g,p,-1000:1000)
end
```

```julia
# Table 5.6 (X^3-3X-t with generic Galois group S_3)

g=X^3-3*X-t
ps=[2,5,7,37,101]

for p in ps 
	println("\n For g=$(g) in the range [-1000,1000] the output of the function rank_with_varying_field for p=$(p) gives the following output:")
	@show rank_with_varying_field(g,p,-1000:1000)
end
```

```julia
# Table 5.7 (X^4+t*X^2+1 with generic Galois group C_2 x C_2)

g=X^4+t*X^2+1

ps=[3,5,7,11,13]

for p in ps 
	println("\n For g=$(g) in the range [-2000,-1] the output of the function rank_with_varying_field for p=$(p) gives the following output:")
	@show rank_with_varying_field(g,p,-2000:-1)
end

for p in ps 
	println("\n For g=$(g) in the range [1,2000] the output of the function rank_with_varying_field for p=$(p) gives the following output:")
	@show rank_with_varying_field(g,p,1:2000)
end
```
