# Chapter 4 - Tame Pro-$p$ Galois Groups of Number Fields
The following code reproduces the linking diagrams presented in Chapter 4 of the thesis. Since the concrete linking diagrams depend on choices, the exact form might differ from the form presented in the document, but they are always equivalent. 

```julia
using Oscar
include("rational_linking_diagrams.jl")
include("general_linking_diagrams.jl")

Qx,x=polynomial_ring(QQ,"x")
empty_ideal_list=Vector{AbsSimpleNumFieldOrderIdeal}()
```

# 4.2 The Cohomological Dimension of the Groups $G_{K,S}^T$
The linking diagrams from Examples 4.2.2 and 4.2.3 can be verified in the following code. 

```julia
# Example 4.2.2

# We start with S={7,13,19,37} then the two additional primes 
# are 10639 and 149713. They correspond to c_1 and c_2 from
# the proof of Theorem 4.2.1 
create_linking_diagram([7,13,19,37,10639, 149713],3)
```

```julia
#Example 4.2.3 (1)
K,a=number_field(x^2-3)
OK=ring_of_integers(K)
a=(OK)(a)

# The set S is given as follows
S=[7*OK,(-a+4)*OK]
# A Koch set S_0 is given as
S0=[5*OK]
@show verify_Koch_set(K,S0,empty_ideal_list,3)

# The set S_1 is determined as follows, where the first two elements correspond 
# to c_1 and c_2 from the proof of Theorem 4.2.1.
S1=[1063*OK,593*OK,(531*a+2224)*OK]


create_linking_diagram_of_K(K,3,union(S,S1),S0,empty_ideal_list)
```

```julia
#Example 4.2.3 (2)
K,a=number_field(x^2-79)
OK=ring_of_integers(K)
a=(OK)(a)

# The set S is given as follows
S=[ideal(OK,7,a+4),ideal(OK,13,a+1)]
# A generating set of cl(K)/p is given by
T=[ideal(OK,3,a+1)]
# A Koch set S_0 is given as
S0=[ideal(OK,7,a+3),11*OK]
@show verify_Koch_set(K,S0,T,3)
# The set S_1 is determined as follows, where the first two elements correspond 
# to c_1 and c_2 from the proof of Theorem 4.2.1.
S1=[ideal(OK,65*a+66),ideal(OK,17389,a+4273),ideal(OK,-2919*a+26594),ideal(OK,86121199, a + 5709782)]


create_linking_diagram_of_K(K,3,union(S,S1),S0,T)
```

# 4.3 Realization of Cohomology Algebras
The following code verifies the results presented in Example 4.3.12 and 4.3.13. Using the function "realization_of_inertial_graph($K$,$p$,$\Gamma$,$S$,$T$)" allows to construct own examples of prime tuples that realize cohomology algebras given by $K$-inertial graphs $\Gamma$. For computational reasons it is only possible to use $K$-inertial graphs for which $|E|=d'_K+|V|$.

```julia
# Example 4.3.12 
K,a=number_field(x^2+31)
OK=ring_of_integers(K)
a=(OK)(a)

T=[ideal(OK,2,(a+3)/2)]
S0=[ideal(OK,7,a+2)]

@show verify_Koch_set(K,S0,T,3)

S=[ideal(OK,7,a+5),ideal(OK,13),ideal(OK,4129),ideal(OK,3613),ideal(OK,257443,a+245436),ideal(OK,1062643,a+778479)]

create_linking_diagram_of_K(K,3,S,S0,T)
```

```julia
# Example 4.3.13
K,a=number_field(x^3-2)
OK=ring_of_integers(K)
a=(OK)(a)

S0=[ideal(OK,a^2+3)]

@show verify_Koch_set(K,S0,empty_ideal_list,3)

S=[ideal(OK,-a^2+2*a+1),ideal(OK,19),ideal(OK,-a^2+10*a+3),ideal(OK,11*a^2-35*a-17),ideal(OK,29*a^2-13*a-5)]

create_linking_diagram_of_K(K,3,S,S0,empty_ideal_list)
```
